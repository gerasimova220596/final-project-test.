```python
print('Пропуски:')
display(ab.isna().sum())

print('Повторы пользователя внутри эксперимента:')
print(ab.duplicated(['experiment_num', 'user_id']).sum())

print('Количество пользователей по группам:')
display(
    ab.groupby(['experiment_num', 'experiment_group'])
    .agg(
        rows=('user_id', 'size'),
        users=('user_id', 'nunique')
    )
)
```

```python
results = []

for experiment in sorted(ab['experiment_num'].unique()):
    data = ab[ab['experiment_num'] == experiment]

    control = data.loc[
        data['experiment_group'] == 'control', 'revenue'
    ].dropna()

    test = data.loc[
        data['experiment_group'] == 'test', 'revenue'
    ].dropna()

    t_stat, p_value = stats.ttest_ind(
        test,
        control,
        equal_var=False,
        alternative='two-sided'
    )

    control_arpu = control.mean()
    test_arpu = test.mean()

    uplift = (
        (test_arpu / control_arpu - 1) * 100
        if control_arpu != 0 else np.nan
    )

    results.append({
        'Эксперимент': int(experiment),
        'Пользователи control': len(control),
        'Пользователи test': len(test),
        'ARPU control': control_arpu,
        'ARPU test': test_arpu,
        'Изменение ARPU, %': uplift,
        'p-value': p_value
    })

ab_results = pd.DataFrame(results)

display(
    ab_results.style.format({
        'ARPU control': '{:.2f}',
        'ARPU test': '{:.2f}',
        'Изменение ARPU, %': '{:.2f}',
        'p-value': '{:.6g}'
    })
)
```

```python
for _, row in ab_results.iterrows():
    experiment = int(row['Эксперимент'])

    if row['p-value'] < 0.05:
        direction = (
            'выше'
            if row['ARPU test'] > row['ARPU control']
            else 'ниже'
        )
        conclusion = f'ARPU тестовой группы статистически значимо {direction}.'
    else:
        conclusion = 'Статистически значимого различия ARPU не обнаружено.'

    print(f'Эксперимент {experiment}: {conclusion}')
```
